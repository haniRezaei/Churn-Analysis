```python
# =========================
# DATA CLEANING
# =========================

# fill missing values
df["Value_Deal"]             = df["Value_Deal"].fillna("None")
df["Multiple_Lines"]         = df["Multiple_Lines"].fillna("No")
df["Internet_Type"]          = df["Internet_Type"].fillna("None")
df["Online_Security"]        = df["Online_Security"].fillna("No")
df["Online_Backup"]          = df["Online_Backup"].fillna("No")
df["Device_Protection_Plan"] = df["Device_Protection_Plan"].fillna("No")
df["Premium_Support"]        = df["Premium_Support"].fillna("No")
df["Streaming_TV"]           = df["Streaming_TV"].fillna("No")
df["Streaming_Movies"]       = df["Streaming_Movies"].fillna("No")
df["Streaming_Music"]        = df["Streaming_Music"].fillna("No")
df["Unlimited_Data"]         = df["Unlimited_Data"].fillna("No")
df["Churn_Category"]         = df["Churn_Category"].fillna("Others")
df["Churn_Reason"]           = df["Churn_Reason"].fillna("Others")
```

```python
print("\nNull Values After Cleaning:")
print(df.isnull().sum())
```

# FEATURE ENGINEERING

```python
# Monthly Charge Range

def monthly_charge_range(charge):
    if charge < 20:    return "< 20"
    elif charge < 50:  return "20-50"
    elif charge < 100: return "50-100"
    else:              return "> 100"
    
df["Monthly_Charge_Range"] = df["Monthly_Charge"].apply(monthly_charge_range)
```

```python
# Age Group

def age_group(age):
    if age < 20:   return "< 20"
    elif age < 36: return "20-35"
    elif age < 51: return "36-50"
    else:          return "> 50"

df["Age_Group"] = df["Age"].apply(age_group)
df["AgeGrp_Sorting"] = df["Age_Group"].map({"< 20": 1, "20-35": 2, "36-50": 3, "> 50": 4})
```

```python
# Tenure Group

def tenure_group(tenure):
    if tenure < 6:    return "< 6 Months"
    elif tenure < 12: return "6-12 Months"
    elif tenure < 18: return "12-18 Months"
    elif tenure < 24: return "18-24 Months"
    else:             return ">= 24 Months"

df["Tenure_Group"] = df["Tenure_in_Months"].apply(tenure_group)
df["TenureGrp_Sorting"] = df["Tenure_Group"].map({"< 6 Months": 1, 
                                                  "6-12 Months": 2,
                                                  "12-18 Months": 3, 
                                                  "18-24 Months": 4, 
                                                  ">= 24 Months": 5})
```

```python
sql_main_dataset = df.copy()
```

# now we create two data set that we will use later in machine learning before making churn status

```python
#making twoو datasets based on "Churned", "Stayed" and the other based on Customer_Status == "Joined"

churn_data = df[df["Customer_Status"].isin(["Stayed", "Churned"])].copy()
join_data  = df[df["Customer_Status"] == "Joined"].copy()
```

```python
# Create Churn_Status only for churn_data
# because join_data customers have not yet been classified as Churned or Stayed

churn_data["Churn_Status"] = churn_data["Customer_Status"].apply(
    lambda x: 1 if x == "Churned" else 0
)

print("\nChurn Data Shape:", churn_data.shape)
print("Join Data Shape:",   join_data.shape)
print("\nChurn Status Values:")
print(churn_data["Churn_Status"].value_counts())
```

```python
# =========================
# Create the Services table
# =========================

# List of service-related columns
service_columns = [
    "Phone_Service", "Multiple_Lines", "Internet_Service",
    "Internet_Type", "Online_Security", "Online_Backup",
    "Device_Protection_Plan", "Premium_Support", "Streaming_TV",
    "Streaming_Movies", "Streaming_Music", "Unlimited_Data"
]

# Convert service columns from wide format to long format
prod_services = churn_data[["Customer_ID"] + service_columns].melt(id_vars="Customer_ID",
                                                                   var_name="Services",
                                                                   value_name="Status")
```

```python
# =========================
# Save the processed datasets 
# =========================

# Save the historical customer data (Stayed + Churned)
churn_data.to_csv("C:/Users/Hanieh/Desktop/churn_data.csv", index=False)

# Save the new customers (Joined)
join_data.to_csv("C:/Users/Hanieh/Desktop/join_data.csv", index=False)

# Save the unpivoted services table
prod_services.to_csv("C:/Users/Hanieh/Desktop/prod_services.csv", index=False)

# Save main table for sql
sql_main_dataset.to_csv(r"C:/Users/Hanieh/Desktop/prod_customers.csv", index=False)


# Display a success message
print("\n files saved successfully!")
```
